# Quantum Teleportation

Quantum teleportation uses a shared entangled state (sometimes called an e-bit) to transfer an unknown quantum state to a distant receiver using classical communication.



## Universal Gate

The **U gate** (Universal Gate) in Qiskit is defined by three parameters: **θ, ϕ, λ**. Its unitary matrix form is:

$$
U(\theta,\phi,\lambda)=
\begin{pmatrix}
\cos\frac{\theta}{2} & -e^{i\lambda}\sin\frac{\theta}{2} \\
e^{i\phi}\sin\frac{\theta}{2} & e^{i(\phi+\lambda)}\cos\frac{\theta}{2}
\end{pmatrix}
$$

When applied to the  state $|0\rangle$, it simplifies to:

$$
|\psi\rangle =
\cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle
$$





## Unknown Quantum State

An unknown quantum state is written as:

$$
|\psi\rangle = \alpha|0\rangle + \beta|1\rangle
$$

where **α** and **β** are complex numbers such that:

$$
|\alpha|^2 + |\beta|^2 = 1
$$

This can also be written as:

$$
|\psi\rangle =
\cos\frac{\theta}{2}|0\rangle + e^{i\phi}\sin\frac{\theta}{2}|1\rangle
$$

or in vector form:

$$
|\psi\rangle =
\begin{pmatrix}
\cos\frac{\theta}{2} \\
e^{i\phi}\sin\frac{\theta}{2}
\end{pmatrix}
$$

### $|\psi\rangle_Q$ State Unknown Quantum State

The secret state is prepared on qubit Q:

$$
|\psi\rangle_Q = \alpha|0\rangle_Q + \beta|1\rangle_Q
$$

where $\alpha$ and $\beta$ are unknown complex amplitudes satisfying $|\alpha|^2 + |\beta|^2 = 1$.



### Alice and Bob   Entangled State

Initially, Alice's qubit and Bob's qubit are each, separately, in the $|0\rangle$ state written together as $|0\rangle_A|0\rangle_B$ or simply as $|00\rangle$.

Alice applies a Hadamard gate on her qubit (first qubit), then a CNOT gate with Alice's qubit as the control and Bob's qubit (second qubit) as the target:

$$
CNOT(A,B)\, H_A |0\rangle_A |0\rangle_B
\;=\; CNOT(A,B)\frac{1}{\sqrt{2}}\bigl(|0\rangle_A + |1\rangle_A\bigr)|0\rangle_B
$$

$$
=\; \frac{1}{\sqrt{2}}\bigl(CNOT(A,B)|0\rangle_A|0\rangle_B + CNOT(A,B)|1\rangle_A|0\rangle_B\bigr)
$$

$$
= \frac{1}{\sqrt{2}}\bigl(|0\rangle_A|0\rangle_B + |1\rangle_A|1\rangle_B\bigr)
$$

Note that now Alice's and Bob's qubits are entangled. Although it is not yet determined by nature whether both their qubits are in the $|0\rangle$ state or the $|1\rangle$ state, it is known that their qubits are in the same state.







### Combined Three Qubit State

At this point Q is simply adjacent to Alice's qubit (A). There has been no entanglement between Q and the pair, so the quantum state of the three qubits together can be written as:

$$
|\psi\rangle_Q|\psi\rangle_{AB} = \bigl(\alpha|0\rangle_Q + \beta|1\rangle_Q\bigr)\frac{1}{\sqrt{2}}\bigl(|0\rangle_A|0\rangle_B + |1\rangle_A|1\rangle_B\bigr)
$$







Alice now operates with a CNOT gate with Q as the control and Alice's qubit as the target, then applies a Hadamard gate to Q. Let's calculate the three-qubit state after that operation:

$$
H_Q\,CNOT(Q,A)|\psi\rangle_Q|\psi\rangle_{AB}
$$

$$
=\; H_Q\,CNOT(Q,A)\bigl(\alpha|0\rangle_Q + \beta|1\rangle_Q\bigr)\frac{1}{\sqrt{2}}\bigl(|0\rangle_A|0\rangle_B + |1\rangle_A|1\rangle_B\bigr)
$$

After applying $CNOT(Q,A)$  when $Q=|0\rangle$, Alice's qubit is unchanged; when $Q=|1\rangle$, Alice's qubit flips:

$$
=\; H_Q\frac{1}{\sqrt{2}}\Bigl[
\alpha|0\rangle_Q\bigl(|0\rangle_A|0\rangle_B + |1\rangle_A|1\rangle_B\bigr)
+ \beta|1\rangle_Q\bigl(|1\rangle_A|0\rangle_B + |0\rangle_A|1\rangle_B\bigr)
\Bigr]
$$

Now applying $H_Q$  using $H|0\rangle = \frac{1}{\sqrt{2}}(|0\rangle+|1\rangle)$ and $H|1\rangle = \frac{1}{\sqrt{2}}(|0\rangle-|1\rangle)$:

$$
=\; \frac{1}{2}\Bigl[
\alpha\bigl(|0\rangle_Q+|1\rangle_Q\bigr)\bigl(|0\rangle_A|0\rangle_B + |1\rangle_A|1\rangle_B\bigr)
+ \beta\bigl(|0\rangle_Q-|1\rangle_Q\bigr)\bigl(|1\rangle_A|0\rangle_B + |0\rangle_A|1\rangle_B\bigr)
\Bigr]
$$

$$
=\; \frac{1}{2}\Bigl[
\alpha|0\rangle_Q|0\rangle_A|0\rangle_B + \alpha|0\rangle_Q|1\rangle_A|1\rangle_B
+ \alpha|1\rangle_Q|0\rangle_A|0\rangle_B + \alpha|1\rangle_Q|1\rangle_A|1\rangle_B
+ \beta|0\rangle_Q|1\rangle_A|0\rangle_B + \beta|0\rangle_Q|0\rangle_A|1\rangle_B
- \beta|1\rangle_Q|1\rangle_A|0\rangle_B - \beta|1\rangle_Q|0\rangle_A|1\rangle_B
\Bigr]
$$



### Grouping by Alice's Measurement Outcomes

Because A and Q are in the same location, we group the terms above according to the outcomes of measurements on qubits Q and A:

$$
|\psi\rangle = \frac{1}{2}\Bigl[
\bigl(\alpha|0\rangle_B + \beta|1\rangle_B\bigr)|0\rangle_Q|0\rangle_A
+ \bigl(\alpha|0\rangle_B - \beta|1\rangle_B\bigr)|1\rangle_Q|0\rangle_A
+ \bigl(\beta|0\rangle_B + \alpha|1\rangle_B\bigr)|0\rangle_Q|1\rangle_A
+ \bigl(-\beta|0\rangle_B + \alpha|1\rangle_B\bigr)|1\rangle_Q|1\rangle_A
\Bigr]
$$



### Alice Measures Qubits Q and A

Now, Alice can measure qubits Q and A. She cannot control the outcome of that measurement, since quantum measurements are probabilistic. So when she measures, there are 4 possible outcomes and all 4 are equally likely:

$$
|0\rangle_Q|0\rangle_A, \quad |1\rangle_Q|0\rangle_A, \quad |0\rangle_Q|1\rangle_A, \quad |1\rangle_Q|1\rangle_A
$$

Each outcome has a different implication for Bob's qubit. 

| Alice Outcome | Bob's State | Instruction to Bob | Result |
|---|---|---|---|
| $\|0\rangle_Q\|0\rangle_A$ | $\alpha\|0\rangle_B + \beta\|1\rangle_B$ | None | $\alpha\|0\rangle_B + \beta\|1\rangle_B$ |
| $\|1\rangle_Q\|0\rangle_A$ | $\alpha\|0\rangle_B - \beta\|1\rangle_B$ | $Z$ | $\alpha\|0\rangle_B + \beta\|1\rangle_B$ |
| $\|0\rangle_Q\|1\rangle_A$ | $\beta\|0\rangle_B + \alpha\|1\rangle_B$ | $X$ | $\alpha\|0\rangle_B + \beta\|1\rangle_B$ |
| $\|1\rangle_Q\|1\rangle_A$ | $-\beta\|0\rangle_B + \alpha\|1\rangle_B$ | $X$ then $Z$ | $\alpha\|0\rangle_B + \beta\|1\rangle_B$ |

## Quantum Teleportation using a Simulator

 To describe the scenario above, we need a circuit with three qubits:  
 two for the entangled pair shared by Alice and Bob, and one for the unknown quantum state |ψ⟩.

```python
from qiskit import QuantumCircuit
import numpy as np
# create 3-qubits circuit
qc = QuantumCircuit(3, 3)

qc.draw(output="mpl")
```

At the start, Alice has an unknown quantum state $\lvert \psi \rangle$.  
We will create this using the $U$ gate.

```python

# create a random 1-qubit state from a random (theta, varphi) 
np.random.seed(1)  # fixing seed for repeatibility

theta = np.random.uniform(0.0, 1.0) * np.pi  # from 0 to pi
varphi = np.random.uniform(0.0, 2.0) * np.pi  # from 0 to 2*pi

# Create the unknown quantum state using the u-gate. Alice has this.

qc.u(theta, varphi, 0.0, 0)
qc.barrier()  # for visual separation

qc.draw(output="mpl")
```

Alice and Bob have a shared entangled pair

```python
# Alice and Bob are together in the same place and set up an entangled pair.
qc.h(1)
qc.cx(1, 2)
qc.barrier()  # for visual separation.
# We can consider that Alice and Bob might move their qubits to different physical locations, now.

qc.draw(output="mpl")
```

Next, Alice entangles the state $\lvert \psi \rangle$ with her part of the shared e-bit using the $CX$ gate and the Hadamard ($H$) gate. She then measures the qubits in the computational basis.

```python
# Alice entangles the unknown state with her part of the e-bit, using the CNOT gate and H gate.
qc.cx(0, 1)
qc.h(0)
qc.barrier()

# Alice measures the two qubits.
qc.measure(0, 0)
qc.measure(1, 1)

qc.draw(output="mpl")
```

Alice sends Bob her measurement results (either "00", "01", "10", or "11").  
Based on these two classical bits, Bob applies a corresponding correction operator on his part of the shared e-bit.  

After this operation, Bob's qubit becomes $\lvert \psi \rangle$.

```python
# Alice sent the results to Bob. Bob applies correction
with qc.if_test((0, 1)):
    qc.z(2)
with qc.if_test((1, 1)):
    qc.x(2)
qc.barrier()

qc.draw(output="mpl")
```

To verify whether the quantum state has been teleported correctly, we apply the inverse of the unitary operator $U$ on Bob's qubit and then measure it in the computational basis.

Since $U^{-1}U = I$ (identity), if Bob's qubit is in the state $U \lvert 0 \rangle$, then applying the inverse operation gives:

$$
U^{-1} U \lvert 0 \rangle = \lvert 0 \rangle
$$

Thus, measuring the qubit should yield the outcome "0", confirming successful teleportation.

```python
# Apply the inverse of u-gate to measure |0>
qc.u(theta, varphi, 0.0, 2).inverse()  # inverse of u(theta,varphi,0.0)
qc.measure(2, 2)  # add measurement gate

qc.draw(output="mpl")
```

```python
from qiskit_aer import AerSimulator
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
from qiskit_ibm_runtime import Sampler
from qiskit.visualization import plot_histogram
from qiskit_aer import StatevectorSimulator



# Define backend
backend = AerSimulator()
nshots = 1000 

# Transpile to backend
pm = generate_preset_pass_manager(backend=backend, optimization_level=1)
isa_qc = pm.run(qc)

# Run the job
sampler = Sampler(mode=backend)
job = sampler.run([isa_qc], shots=nshots)
result = job.result()

# Extract counts data
counts = result[0].data.c.get_counts()
print(counts)

# Plot the counts in a histogram
plot_histogram(counts)
```

## Teleportation on a real quantum computer

```python
# Load the Qiskit Runtime service
from qiskit_ibm_runtime import QiskitRuntimeService

# Load the Qiskit Runtime service

# Syntax for first saving your token.  Delete these lines after saving your credentials.
# QiskitRuntimeService.save_account(channel='ibm_quantum_platform', instance = '<YOUR_IBM_INSTANCE_CRN>', token='<YOUR-API_KEY>', overwrite=True, set_as_default=True)
# service = QiskitRuntimeService(channel='ibm_quantum_platform')

# Load saved credentials
service = QiskitRuntimeService()

# Use the least busy backend, or uncomment the loading of a specific backend like "ibm_brisbane".
backend = service.least_busy(operational=True, simulator=False, min_num_qubits=127)
# backend = service.backend("ibm_brisbane")
print(backend.name)
```

```python
#  Transpile
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager

target = backend.target
pm = generate_preset_pass_manager(target=target, optimization_level=3)
qc_isa = pm.run(qc)

```

```python
# Load the Runtime primitive and session
from qiskit_ibm_runtime import SamplerV2 as Sampler

sampler = Sampler(mode=backend)
# Load the backend sampler
from qiskit.primitives import BackendSamplerV2

# Load the Aer simulator and generate a noise model based on the currently-selected backend.
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel


noise_model = NoiseModel.from_backend(backend)

# Define a simulator using Aer, and use it in Sampler.
backend_sim = AerSimulator(noise_model=noise_model)
sampler_sim = BackendSamplerV2(backend=backend_sim)

```

```python
job = sampler.run([qc_isa])
# job = sampler_sim.run([qc_isa])
res = job.result()
counts = res[0].data.c.get_counts()


# This required 5 s to run on a Heron r2 processor on 10-28-24
from qiskit.visualization import plot_histogram

plot_histogram(counts)
```
